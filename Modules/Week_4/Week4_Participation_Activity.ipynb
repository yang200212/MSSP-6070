```python
# Find the 95th percentile of earthquake magnitude in Japan using magType of 'mb'
japan_mb = parsed[
    (parsed["parsed_place"] == "Japan") &
    (parsed["magType"] == "mb")
]

japan_mb["mag"].quantile(0.95)
```

```python
# Find the percentage of earthquakes in Indonesia that were coupled with tsunamis

# Establish new dataset called 'indo' by filtering out Indonesia rows
indo = parsed[parsed["parsed_place"] == "Indonesia"]

# Count tsunami = 1 and tsunami = 0
tsunami_counts = indo["tsunami"].value_counts()

tsunami_1 = tsunami_counts[1]
tsunami_0 = tsunami_counts[0]

# Calculate percentage of earthquakes coupled with tsunamis
tsunami_percentage = tsunami_1 / (tsunami_1 + tsunami_0) * 100
tsunami_percentage
```

```python
# Get summary statistics for earthquakes in Nevada
nevada = parsed[parsed["parsed_place"] == "Nevada"]
nevada.describe()
```

```python
# Add a column to the dataframe indicating whether or not the earthquake happened in a country or US state that is on the Ring of Fire.
# Use Bolivia, Chile, Ecuador, Peru, Costa Rica, Guatemala, Mexico (be sure not to select New Mexico), Japan, Philippines,
# Indonesia, New Zealand, Antarctica (look for the Antarctic), Canada, Fiji, Alaska, Washington, California, Russia, Taiwan, Tonga, and the Kermadec Islands)

# Establish ring of fire locations
ring_locations = [
    "Bolivia",
    "Chile",
    "Ecuador",
    "Peru",
    "Costa Rica",
    "Guatemala",
    "Mexico",
    "Japan",
    "Philippines",
    "Indonesia",
    "New Zealand",
    "Canada",
    "Fiji",
    "Alaska",
    "Washington",
    "California",
    "Russia",
    "Taiwan",
    "Tonga",
    "Kermadec Islands"
]

# Create new variables called 'ring_of_fire'
parsed["ring_of_fire"] = (
    parsed["parsed_place"].isin(ring_locations) |
    parsed["parsed_place"].str.contains("Antarctic", case=False, na=False)
)

# check
parsed["ring_of_fire"]
```

```python
# Calculate the number of earthquakes in the Ring of Fire locations and the number outside of them
parsed["ring_of_fire"].value_counts()

# number of earthquakes in the Ring of Fire: 7188
# outside: 2144
```

```python
# Find the tsunami count along the Ring of Fire
ring_tsunamis = parsed[
    (parsed["ring_of_fire"] == True) &
    (parsed["tsunami"] == 1)
]

len(ring_tsunamis) # 45 tsunamis along the Ring of Fire
```

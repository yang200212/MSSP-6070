Answer the following questions:

a. When does the code in a function execute: when the function is defined or called?

b. How many global scopes are there in a Python program? How many local scopes?

c. What happens to variables in a local scope when the function call returns?

d. How can you prevent a program from crashing when it gets an error?

a. The code inside a function executes when the function is called, not when it is defined. Defining a function using "def" simply tells Python what the function should do, while calling it actually runs the instructions.

b. A Python program typically has one global scope. It can have multiple local scopes because each function call creates its own local scope.

c. When a function call returns, its local scope is no longer active, and its local variables are no longer accessible. However, values returned by the function can be stored in variables outside the function and used later.

d. I can use 'try' and 'except' statements to handle errors. Python would attempt to execute the code in the try block, and if a matching exception occurs, the except block handles it instead of allowing the program to crash.

```python
# Define a function named collatz() that has one parameter named number
def collatz(number):
  if number % 2 == 0:
    result = number // 2
    print(result)
    return result
  else:
    result = 3 * number + 1
    print(result)
    return result
```

```python
number = int(input("Enter number: "))

while number != 1:
    number = collatz(number)
```

### sample variance
Remember to use Bessel's correction.

I will use this formula to calculate the sample variance of salaries:
$$s^2 = \frac{\sum_{1}^{n} (x_i - \bar{x})^2}{n - 1}$$

```python
statistics.variance(salaries)
```

### sample standard deviation
Remember to use Bessel's correction.

I will use this formula to calculate the sample standard deviation of salaries:
$$s = \sqrt{\frac{\sum_{1}^{n} (x_i - \bar{x})^2}{n - 1}} = \sqrt{s^2}$$

```python
s = statistics.stdev(salaries)
s
```

## Exercise 7: Scaling data
### min-max scaling

```python
# By using min-max scaling, I can convert the original units of all values from 'salaries' into a common scale.
min = min(salaries)
max = max(salaries)
salaries_scaled = [(value - min) / (max - min) for value in salaries]
print(salaries_scaled)
```

### standardizing

```python
# I can calculate z-score to standardize all data from 'salaries'.
z_score = [(value - x) / s for value in salaries]
z_score
```

```python
from google.colab import userdata
import os
import subprocess
from pathlib import Path

owner = "yang200212" # Replace with the GitHub repository owner
repository = "MSSP-6070" # Replace with the GitHub repository name
repo_url = f'https://github.com/{owner}/{repository}.git'
repo_path = Path('/content') / repository

# Clone the repository
import subprocess
subprocess.run(['git', 'clone', repo_url])

# Navigate into the cloned repository directory (optional, but often useful)
os.chdir(repository)
print("Changed directory to: {os.getcwd()}")
```

```python
import pandas as pd

# Define dataset path
DATA_PATH = repo_path / 'Modules' / 'Week_6' / 'Student Performance Data.csv'

# Check whether the file exists
if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"Data file not found: {DATA_PATH}"
    )

# Load the CSV dataset
student = pd.read_csv(DATA_PATH, encoding='cp1252')

# Confirm successful loading
print("Data loaded successfully!")
print("Data shape:", student.shape)
display(student.head())
```

```python
# take a look at 'student' dataset's variables
student.head()
```

```python
# write a lambda function that transforms 'gender' to a binary variable, with female=0 and male=1
student_transform = lambda df: df.assign(
    gender_new=df["gender"].apply(lambda x: 0 if x == "female" else 1)
)

student_new = student_transform(student)
```

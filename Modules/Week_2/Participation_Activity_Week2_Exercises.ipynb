3(c) letters[-1] also evaluates to 'd' because [-1] means the last element in the list. Negative indexes count backward from the end of a list.

3(d) letters[:2] evaluates to ['a', 'b'] because [:2] means taking everything from the start of the list to index 2, but excluding index 2. Excluding letters[2], which is 'c', we get ['a', 'b'].

5.Write a function that accepts the list 'spam' as an argument and returns a string with all the items separated by a comma and a space, with 'and' inserted before the last item.

```python
spam_new = ['apples', 'bananas', 'tofu', 'cats']

def list_to_string(items):
    if len(items) == 0:
        return ''
    elif len(items) == 1:
        return str(items[0])
    else:
        return ', '.join(str(item) for item in items[:-1]) + ', and ' + str(items[-1])

print(list_to_string(spam_new))

# str(items) just in case the list contains values other than strings, for example, numbers/other non-string values.
# items[:-1] gets everything except the last item,
# ', '.join(...) combines them,
# items[-1] gets the last item,
# then add ', and ' before the last item.
```

```python
# add a column to each dataframe called 'ticker', indicating the ticker symbol it is for (Apple is AAPL)
facebook["ticker"] = "FB"
apple["ticker"] = "AAPL"
amazon["ticker"] = "AMZN"
netflix["ticker"] = "NFLX"
google["ticker"] = "GOOG"
```

```python
# append each file together into a single dataframe 'faanf'
faanf = pd.concat(
    [facebook, apple, amazon, netflix, google],
    ignore_index=True
)
```

```python
# save faanf into csv
faanf.to_csv("faanf.csv", index=False)
```

```python
faanf = pd.read_csv("faanf.csv")
# use type conversion to change the 'date' column into a 'datetime' and the 'volume' column into integers
faanf["date"] = pd.to_datetime(faanf["date"])
faanf["volume"] = faanf["volume"].astype(int)

# sort by 'date' and 'ticker'
faanf = faanf.sort_values(by = ["date", "ticker"])
```

```python
faanf.head()
```

```python
# Find the seven rows within the highest value for volume
faanf.nlargest(7, "volume")
```

```python
# Use the melt() method to make the 'date' and 'ticker' key identifiers
# Melt the remaining columns so that they are not separated
faanf_melted = faanf.melt(
    id_vars=["date", "ticker"]
)

faanf_melted.head()
```
